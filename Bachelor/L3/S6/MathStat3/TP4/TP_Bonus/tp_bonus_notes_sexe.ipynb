## Introduction

Ce TP bonus explore les **notes d'algèbre** sur trois années (2022-2024) pour analyser s'il existe une différence significative entre les résultats selon le **sexe** (M/F).

Outils utilisés : statistiques descriptives, tests de normalité, ANOVA, tests non paramétriques.

---
## 1. Importation des données

```r
alg22 = read.csv2("alg22.csv")
alg23 = read.csv2("alg23.csv")
alg24 = read.csv2("alg24.csv")
head(alg22)
head(alg23)
head(alg24)
```

---
## 2. Analyse par année

### 2.1 Année 2022

```r
note22 = alg22$note22
sexe22 = alg22$Sexe

cat("Moyenne F :", mean(note22[sexe22 == "F"], na.rm = TRUE))
cat("\nMoyenne M :", mean(note22[sexe22 == "M"], na.rm = TRUE))
```

```r
boxplot(note22 ~ sexe22, main = "Notes algèbre 2022",
        col = c("pink", "lightblue"), ylab = "Note")
```

```r
# Normalité par groupe
tapply(note22, sexe22, shapiro.test)
```

```r
# ANOVA
anova(lm(note22 ~ sexe22))
# Wilcoxon (non paramétrique)
wilcox.test(note22 ~ sexe22)
```

### 2.2 Année 2023

```r
note23 = alg23$note23
sexe23 = alg23$Sexe

cat("Moyenne F :", mean(note23[sexe23 == "F"], na.rm = TRUE))
cat("\nMoyenne M :", mean(note23[sexe23 == "M"], na.rm = TRUE))
```

```r
boxplot(note23 ~ sexe23, main = "Notes algèbre 2023",
        col = c("pink", "lightblue"), ylab = "Note")
```

```r
tapply(note23, sexe23, shapiro.test)
anova(lm(note23 ~ sexe23))
wilcox.test(note23 ~ sexe23)
```

### 2.3 Année 2024

```r
note24 = alg24$note24
sexe24 = alg24$Sexe

cat("Moyenne F :", mean(note24[sexe24 == "F"], na.rm = TRUE))
cat("\nMoyenne M :", mean(note24[sexe24 == "M"], na.rm = TRUE))
```

```r
boxplot(note24 ~ sexe24, main = "Notes algèbre 2024",
        col = c("pink", "lightblue"), ylab = "Note")
```

```r
tapply(note24, sexe24, shapiro.test)
anova(lm(note24 ~ sexe24))
wilcox.test(note24 ~ sexe24)
```

---
## 3. Analyse globale (toutes années)

```r
notes_all = c(note22, note23, note24)
sexe_all = factor(c(as.character(sexe22), as.character(sexe23), as.character(sexe24)))
annee_all = factor(c(rep("2022", length(note22)), rep("2023", length(note23)), rep("2024", length(note24))))
```

```r
boxplot(notes_all ~ sexe_all, main = "Notes (toutes années)",
        col = c("pink", "lightblue"), ylab = "Note")
```

```r
# ANOVA à deux facteurs
anova(lm(notes_all ~ sexe_all * annee_all))
```

```r
anova(lm(notes_all ~ sexe_all + annee_all))
```

```r
interaction.plot(annee_all, sexe_all, notes_all,
                 main = "Interaction Année × Sexe",
                 col = c("pink", "lightblue"), lwd = 2,
                 xlab = "Année", ylab = "Moyenne des notes")
```

```r
# Kruskal-Wallis
kruskal.test(notes_all ~ sexe_all)
```

---
## Résumé

- Toujours vérifier la **normalité** avant l'ANOVA
- Le test de **Wilcoxon** est une bonne alternative pour 2 groupes
- L'ANOVA à **deux facteurs** sépare l'effet du sexe, de l'année, et de leur interaction
